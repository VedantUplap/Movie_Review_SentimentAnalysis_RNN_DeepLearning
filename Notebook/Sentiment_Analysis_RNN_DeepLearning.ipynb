```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,SimpleRNN,Dense
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data(num_words = 10000)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
print(X_train[0])
```

```python
X_train = pad_sequences(X_train,padding = 'post', maxlen = 100)
X_test = pad_sequences(X_test,padding = 'post',maxlen = 100)
```

```python
print(X_train.shape)
```

## Building RNN model with Embedding Layer

```python
model = Sequential()
model.add(Embedding(input_dim= 10000,output_dim = 32,input_length = 100))

model.add(SimpleRNN(32))
model.add(Dense(1,activation = 'sigmoid'))
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

## training model

```python
history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

## evaluating model

```python
loss, accuracy = model.evaluate(X_test, y_test)

print("Test Accuracy:", accuracy)
```

```python
model.save("sentiment_rnn_model.h5")
```

```python
word_index = imdb.get_word_index()
```

## function for text preprocessing 

```python
def encode_review(text):
    
    words = text.lower().split()
    
    encoded = []

    for word in words:
        index = word_index.get(word)

        if index is not None and index < 10000:
            encoded.append(index + 3)
    
    padded = pad_sequences([encoded], maxlen=100, padding='post')

    return padded
```

## Text prediction

```python
sample = "this movie was amazing and fantastic"

encoded_sample = encode_review(sample)

prediction = model.predict(encoded_sample)

print(prediction)
```

```python
if prediction[0][0] > 0.5:
    print("Positive Review")
else:
    print("Negative Review")
```
